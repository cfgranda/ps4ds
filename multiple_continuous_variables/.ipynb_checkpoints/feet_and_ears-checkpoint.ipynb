```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
from scipy.stats import gaussian_kde, norm

with open("../data/feet_and_ears.pkl", "rb") as f:
    body_measurements = pickle.load(f)
```

```python
percentile_intervals = {
    'foot': (45, 55),
    'forearm': (45, 55),
    'upperarm': (45, 55),
    'lowerleg': (45, 55),
    'upperleg': (45, 55),
}

first_key = next(iter(percentile_intervals))
simultaneous_mask = np.ones(len(body_measurements[first_key]), dtype=bool)

# 2. Loop through each body part and combine the conditions
for col, (low_p, high_p) in percentile_intervals.items():
    arr = body_measurements[col]

    # Calculate values at lower and upper percentiles
    low_val = np.percentile(arr, low_p)
    high_val = np.percentile(arr, high_p)

    # Combine with current mask using bitwise AND (&)
    in_interval = (arr >= low_val) & (arr <= high_val)
    simultaneous_mask &= in_interval

# 3. Sum up the True values in the final combined mask
simultaneous_count = np.sum(simultaneous_mask)
total_count = len(body_measurements[first_key])

print(
    f"Simultaneous match count: {simultaneous_count} out of {total_count} data points"
)
print(f"Percentage: {(simultaneous_count / total_count) * 100:.2f}%")
```

```python
def compute_conditional_gaussian(data_dict, target_var, given_vars, given_values):
    """Fits a parametric Gaussian random vector model and computes the conditional mean 
    and standard deviation.
    """
    all_vars = given_vars + [target_var]

    # Combine dictionary values into a 2D matrix (N_samples x N_features)
    X = np.column_stack([data_dict[var] for var in all_vars])

    # Compute empirical mean vector and covariance matrix
    mu = np.mean(X, axis=0)
    cov_matrix = np.cov(X, rowvar=False)

    k = len(given_vars)
    mu_x = mu[:k]
    mu_y = mu[k]

    Sigma_xx = cov_matrix[:k, :k]
    Sigma_xy = cov_matrix[:k, k:]
    Sigma_yx = cov_matrix[k:, :k]
    Sigma_yy = cov_matrix[k, k]

    # Conditional multivariate Gaussian distribution formulas
    inv_Sigma_xx = np.linalg.inv(Sigma_xx)
    diff = np.array(given_values) - mu_x

    mu_cond = mu_y + (Sigma_yx @ inv_Sigma_xx @ diff)[0]
    sigma2_cond = Sigma_yy - (Sigma_yx @ inv_Sigma_xx @ Sigma_xy)[0, 0]

    return mu_cond, np.sqrt(sigma2_cond)
```

```python
def plot_conditional_pdfs(
    data_dict, target_var, predictors, percentiles=range(10, 100, 10)
):
    """Computes and plots conditional PDFs for varying 'foot' percentiles
    alongside the marginal KDE density plot.
    """
    # 1. 50th percentile for given predictor variables
    p50_vals = {var: np.percentile(data_dict[var], 50) for var in predictors}

    # 2. Define grid values for target variable pdf evaluation
    target_data = data_dict[target_var]
    x_grid = np.linspace(
        target_data.min() - 0.2 * target_data.std(),
        target_data.max() + 0.2 * target_data.std(),
        500,
    )

    plt.figure(figsize=(10, 5))

    # 3. Compute and plot marginal KDE
    kde = gaussian_kde(target_data)
    plt.plot(
        x_grid,
        kde(x_grid),
        color='black',
        linestyle='--',
        linewidth=2.5,
        label=f'Marginal KDE ({target_var})',
    )

    # 4. Compute conditional pdfs across 'foot' percentiles
    colors = plt.cm.inferno(np.linspace(0, 0.75, len(percentiles)))

    for p, color in zip(percentiles, colors):
        given_values = []
        for var in predictors:
            if var == 'foot':
                given_values.append(np.percentile(data_dict['foot'], p))
            else:
                given_values.append(p50_vals[var])

        mu_cond, std_cond = compute_conditional_gaussian(
            data_dict, target_var, predictors, given_values
        )
        pdf = norm.pdf(x_grid, loc=mu_cond, scale=std_cond)

        plt.plot(
            x_grid,
            pdf,
            color=color,
            linewidth=1.8,
            alpha=0.85,
            label=f'Foot {p}th percentile',
        )

    plt.xlabel(f'{target_var.capitalize()} length', fontsize=11)
    plt.ylabel('Probability Density', fontsize=11)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

```python
# Predictor variables used for conditioning
predictors = ['foot', 'forearm', 'upperarm', 'lowerleg', 'upperleg']


plot_conditional_pdfs(body_measurements, 'hand', predictors)
plot_conditional_pdfs(body_measurements, 'ear', predictors)
```
