Estimation of the average treatment effect of legal access to alcohol on mortality due to motor vehicle accidents for people between 20.5 and 21.5 years of age.\

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = np.load('../data/drinking_age.npz')

# Extract the individual arrays
age = data['age']
mortality = data['mortality']

fig, (ax1) = plt.subplots(1, 1, figsize=(7, 5))

# --- Plot 1: Motor Vehicle Accidents (MVA) ---
ax1.scatter(age, mortality, color='crimson', s=50, alpha=0.8)
ax1.axvline(x=21.0, color='gray', linestyle='--', linewidth=1.5)
ax1.set_xlabel("Age")
ax1.set_ylabel("Deaths per 100,000 person-years (motor vehicle accidents)")
ax1.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# We remove the data point right after the 21st birthday to account for partying as a confounding factor
ATE = np.mean(mortality[age>21.][1:]) - np.mean(mortality[age<21.]) 
print("ATE",np.around(ATE,2),"Deaths per 100,000 person-years (motor vehicle accidents)")
```
