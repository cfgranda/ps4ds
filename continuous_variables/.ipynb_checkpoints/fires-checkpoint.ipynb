Prediction of rare events (catastrophically large fires) using a parametric Pareto distribution.\

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

fontsize = 20
```

```python
cnx = sqlite3.connect('../data/FPA_FOD_20170508.sqlite')
df = pd.read_sql_query("SELECT FIRE_YEAR,FIRE_SIZE FROM 'Fires'", cnx)
print(df.head()) #check the data
```

```python
train_set_year = 1995
# Filter for train_set_year and convert to a NumPy vector
fires_train = df[(df['FIRE_YEAR'] >= 1995) & (df['FIRE_YEAR'] <= train_set_year)]['FIRE_SIZE'].to_numpy()
fires_train[fires_train <= 0.5] = 0.5

# Filter for all years after train_set_year and convert to a NumPy vector
fires_test = df[df['FIRE_YEAR'] > train_set_year]['FIRE_SIZE'].to_numpy()
fires_test[fires_test <= 0.5] = 0.5

# Quick check of the vectors
print(f"train set n: {n_train}")
print(f"test set n: {n_test}")
np.savez("fires.npz", fires_train=fires_train, fires_test=fires_test)
```

```python
# Load the data
data = np.load("../data/fires.npz")

fires_train = data["fires_train"]
fires_test = data["fires_test"]

n_train = len(fires_train)
n_test = len(fires_test)
```

```python
n_train = len(fires_train)
n_test = len(fires_test)

catastrophic_threshold = 1e5
n_fires_pred = 23 * 70000 # Assumed total fires in next 23 years
empirical_prob = np.sum(fires_train > catastrophic_threshold) / n_train
predicted_fires_empirical_prob = n_fires_pred * empirical_prob
print("Predicted catastrophic fires via empirical pmf:",predicted_fires_empirical_prob)
print("True catastrophic fires",np.sum(fires_test > catastrophic_threshold))
```

```python
# Exponential model
lambda_ml = 1.0 / np.mean(fires_train)
prob_exponential = np.exp(-lambda_ml * catastrophic_threshold)
predicted_fires_exponential = n_fires_pred * prob_exponential
print("Predicted catastrophic fires via exponential model:",predicted_fires_exponential)
```

```python
ratio_1 = np.sum(fires_train > 1) / np.sum(fires_train > 10) 
ratio_2 = np.sum(fires_train > 10) / np.sum(fires_train > 100) 
ratio_3 = np.sum(fires_train > 100) / np.sum(fires_train > 1000)  
print(np.around(ratio_1,2),np.around(ratio_2,2),np.around(ratio_3,2))
```

```python
# Fit Pareto model via maximum likelihood estimation
b_ml = np.min(fires_train)
alpha_ml = n_train / (np.sum(np.log(fires_train/b_ml)))
```

```python
x = np.geomspace(b_ml, catastrophic_threshold*10, 1000)

# 2. Compute PDFs
pdf_exp = lambda_ml * np.exp(-lambda_ml * x)
pdf_pareto = alpha_ml * (b_ml**alpha_ml) / (x**(alpha_ml + 1))

# 3. Create log-spaced bins for histogram
bins = np.geomspace(np.min(fires_train), np.max(fires_train), 50)

fig, ax = plt.subplots(1, 1, figsize=(20, 4), sharex=True)

# Plot histogram with log-spaced bins
ax.hist(
    fires_train,
    bins=bins,
    color="skyblue",
    edgecolor="black",
    density=True,
    log=True,
)

# # Plot curves
ax.plot(x, pdf_exp, color="red", linewidth=2, label="Exponential")
ax.plot(x, pdf_pareto, color="purple", linewidth=2, label="Pareto")

# # 4. Set both axes to log scale
ax.set_xscale("log")
# ax.set_yscale("log")
ax.set_ylim([1e-10,10])

ax.set_xlabel("Fire size (acres)", fontsize=fontsize)
ax.set_ylabel("Probability density", fontsize=fontsize)
ax.tick_params(axis="both", labelsize=fontsize)

ax.legend(fontsize=fontsize)

# Adjust layout and display
plt.tight_layout()
```

```python
prob_pareto = (b_ml/catastrophic_threshold)**alpha_ml
predicted_fires_pareto = n_fires_pred * prob_pareto
print("Predicted catastrophic fires via Pareto model:",np.around(predicted_fires_pareto,2))
```

```python
b = np.min(fires_train)
ratio_0 = np.sum(fires_train > 0.01) / np.sum(fires_train > 0.1) 
ratio_1 = np.sum(fires_train > 1) / np.sum(fires_train > 10) 
ratio_2 = np.sum(fires_train > 10) / np.sum(fires_train > 100) 
ratio_3 = np.sum(fires_train > 100) / np.sum(fires_train > 1000) 
ratio_4 = np.sum(fires_train > 1000) / np.sum(fires_train > 10000) 
print(b,ratio_0,ratio_1,ratio_2,ratio_3,ratio_4)
```

```python
b = np.min(fires_test)
ratio_0 = np.sum(fires_test > 0.01) / np.sum(fires_test > 0.1) 
ratio_1 = np.sum(fires_test > 1) / np.sum(fires_test > 10) 
ratio_2 = np.sum(fires_test > 10) / np.sum(fires_test > 100) 
ratio_3 = np.sum(fires_test > 100) / np.sum(fires_test > 1000) 
ratio_4 = np.sum(fires_test > 1000) / np.sum(fires_test > 10000) 
print(b,ratio_0,ratio_1,ratio_2,ratio_3,ratio_4)
```

```python
b = np.min(fires_train)
alpha_ml = n_train / (np.sum(np.log(fires_train/b)))
print(alpha_ml)
```

```python
thresh = 1e5
print("Test probability",np.sum(fires_test > thresh) / n_test )
print("Empirical probability",np.sum(fires_train > thresh) / n_train )
print("Pareto model",(b/thresh)**alpha_ml)
```

```python
shape, loc, scale = stats.pareto.fit(fires_train, floc=0)

# 3. Compute P(X >= 100000)
# The Survival Function (sf) gives P(X >= x)
probability = stats.pareto.sf(thresh, b=shape, loc=loc, scale=scale)

print(f"Fitted Parameters:")
print(f"  Shape (alpha): {shape:.4f}")
print(f"  Scale (xm): {scale:.4f}")
print(f"\nProbability that the variable is >= {thresh}: {probability:.6f}")
print(f"Or as a percentage: {probability * 100:.4f}%")
```

```python
shape, loc, scale = stats.lognorm.fit(fires_train, floc=0)
probability = stats.lognorm.sf(thresh, s=shape, loc=loc, scale=scale)

print(f"Fitted Parameters:")
print(f"  Shape (sigma): {shape:.4f}")
print(f"  Scale (exp(mu)): {scale:.4f}")
print(f"\nProbability that the variable is >= {thresh}: {probability:.10f}")
print(f"Or as a percentage: {probability * 100:.4f}%")
```
